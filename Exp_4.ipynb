# **Logical Operators**

Use of | & ||

```r
x = 8
(x<10) || (x<2)
```

```r
x = 18
(x<10) || (x<2)
```

```r
x = c(8,18)

(x<10) | (x<2)
```

Use of & and &&

```r
x = 5
(x<10) && (x>2)
```

```r
x = c(8,18)
(x<10) & (x>2)
```

```r
x = 1:6
(x>2) & (x<5)
```
